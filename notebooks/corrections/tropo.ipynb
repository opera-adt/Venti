```python

"""Script to crop TROPO products for multiple frames using opera-utils tropo-crop."""

import json
import logging
import subprocess
import sys
from pathlib import Path

import opera_utils

# Set up logging
logging.basicConfig(
    level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s"
)
logger = logging.getLogger(__name__)


def get_bounds_from_geometry(geometry):
    """Extract bounds from shapely geometry as (west, south, east, north)."""
    bounds = geometry.bounds
    return bounds  # Already in (minx, miny, maxx, maxy) format


def run_tropo_crop(
    frame_id,
    bounds,
    datetimes,
    tropo_urls_file="tropo_urls_combined.txt",
    output_dir=None,
    num_workers=5,
):
    """Run opera-utils tropo-crop for a single frame."""
    if output_dir is None:
        output_dir = f"cropped_tropo_frame_{frame_id}"

    # Create output directory if it doesn't exist
    Path(output_dir).mkdir(parents=True, exist_ok=True)

    # Build the command
    cmd = ["opera-utils", "tropo-crop", "--tropo-urls-file", tropo_urls_file, "--aoi-bounds", str(bounds[0]), str(bounds[1]), str(bounds[2]), str(bounds[3]), "--output-dir", output_dir, "--num-workers", str(num_workers), "--datetimes", *datetimes]

    logger.info(f"Running command for frame {frame_id}: {' '.join(cmd)}")

    try:
        subprocess.run(cmd, text=True, check=True)
        logger.info(f"Successfully processed frame {frame_id}")
        return True
    except subprocess.CalledProcessError as e:
        logger.exception(f"Error processing frame {frame_id}:")
        logger.exception(f"Command: {' '.join(cmd)}")
        logger.exception(f"Return code: {e.returncode}")
        logger.exception(f"STDOUT: {e.stdout}")
        logger.exception(f"STDERR: {e.stderr}")
        return False


# for fid in 33039 33038 23210 23211 20696 20697 40296 40297; do opera-utils tropo-apply --dem-path cropped_tropo_frame_${fid}/geometry/dem*tif --incidence-angle-path cropped_tropo_frame_${fid}/geometry/incidence_angle.tif --interp-method cubic --cropped-tropo-list cropped_tropo_frame_${fid}/tropo_cropped*.nc; done
def apply(frame_id: int, dem_path: str | Path) -> Path:
    # TODO:
    # Here was the command line args i ran:
    #  python -m opera_utils.disp.create_incidence_range --los-enu geometry/los_enu.tif
    # opera-utils tropo-apply --dem-path cropped_tropo_frame_${fid}/geometry/dem*tif \
    #     --incidence-angle-path cropped_tropo_frame_${fid}/geometry/incidence_angle.tif \
    #     --interp-method cubic
    #     --cropped-tropo-list cropped_tropo_frame_${fid}/tropo_cropped*.nc
    pass


def main():
    # Configuration
    json_url = "https://github.com/opera-adt/burst_db/releases/download/v0.13.0/opera-disp-s1-consistent-burst-ids-2025-09-16-2016-07-01_to_2024-12-31.json"
    json_file = Path(json_url).name
    import requests

    resp = requests.get(json_url)
    Path(json_file).write_text(resp.text)
    # TODO: get urls from ASF?
    tropo_urls_file = "tropo_urls_combined.txt"
    try:
        num_workers = int(sys.argv[1])
    except:
        num_workers = 5

    # Check if tropo URLs file exists
    if not Path(tropo_urls_file).exists():
        msg = f"TROPO URLs file not found: {tropo_urls_file}"
        raise ValueError(msg)

    # Load JSON data
    logger.info(f"Loading JSON data from {json_file}")
    with open(json_file) as f:
        data = json.load(f)
        # TEST

    # Get frame geodataframe
    logger.info("Loading frame geodataframe...")
    frame_gdf = opera_utils.get_frame_geodataframe()

    # Process each frame
    frame_data = data.get("data", {})
    total_frames = len(frame_data)
    successful = 0
    failed = 0

    logger.info(f"Processing {total_frames} frames...")
    # frames = [33039, 33038, 23210, 23211, 20696, 20697, 40296, 40297]
    frames = [40296, 40297]

    for i, (frame_id_str, frame_info) in enumerate(frame_data.items(), 1):
        if int(frame_id_str) not in frames:
            continue
        logger.info(f"Processing frame {i}/{total_frames}: {frame_id_str}")

        # Convert frame_id to int for geodataframe lookup
        frame_id = int(frame_id_str)

        # Get bounds from geometry
        row = frame_gdf.loc[frame_id]
        geometry = row.geometry
        print(row)
        bounds = get_bounds_from_geometry(geometry)

        # Get sensing times
        sensing_times = frame_info.get("sensing_time_list", [])
        if not sensing_times:
            logger.warning(f"No sensing times found for frame {frame_id}")
            failed += 1
            continue

        # Run tropo-crop for this frame
        output_dir = f"cropped_tropo_frame_{frame_id}"
        success = run_tropo_crop(
            frame_id=frame_id,
            bounds=bounds,
            datetimes=sensing_times,
            tropo_urls_file=tropo_urls_file,
            output_dir=output_dir,
            num_workers=num_workers,
        )

        if success:
            successful += 1
        else:
            failed += 1

    # Summary
    logger.info("=" * 50)
    logger.info("Processing complete!")
    logger.info(f"Total frames: {total_frames}")
    logger.info(f"Successful: {successful}")
    logger.info(f"Failed: {failed}")
    logger.info("=" * 50)


if __name__ == "__main__":
    main()
```

```python
import geopandas as gpd
import requests
from shapely.geometry import shape
```

```python
def load_burst_db_to_gdf(url):
    """Load OPERA burst database JSON to GeoDataFrame."""
    resp = requests.get(url)
    resp.raise_for_status()
    data = resp.json()
    print('Loading OPERA consistent burst_id database')
    print(f'  database generation date: {data['metadata']['generation_time']}')

    # Convert dict to list of features
    features = []
    for burst_id, info in data['data'].items():
        features.append({
            'burst_id': burst_id,
            'geometry': shape(info['geometry']) if 'geometry' in info else None,
            **{k: v for k, v in info.items() if k != 'geometry'}
        })

    return gpd.GeoDataFrame(features, crs='EPSG:4326')
```

```python

import asf_search as asf
import pandas as pd


def get_sensing_time_cslc(
    frame_wkt: str,
    orbit_pass: str,
    round_min: int = 2
) -> pd.DatetimeIndex:
    """Search ASF-DAAC for CSLCs and return unique sensing times for the path
    with the highest average burst count.

    Parameters
    ----------
    frame_wkt : str
        WKT geometry string for spatial search (AOI)
    orbit_pass : str
        Orbit direction ('ASCENDING' or 'DESCENDING')
    round_min : int, default=2
        Minutes to round sensing times to

    Returns
    -------
    pd.DatetimeIndex
        Sorted unique sensing times for the optimal path

    """
    print('Searching ASF-DAAC for existing CSLCs')

    # Search ASF-DAAC
    cslc_results = asf.search(
        intersectsWith=frame_wkt,
        flightDirection=orbit_pass,
        collections=['C2777443834-ASF']
    )

    # Convert to GeoDataFrame
    cslc_df = gpd.GeoDataFrame.from_features(cslc_results.geojson())

    if cslc_df.empty:
        print('No CSLCs found for the specified parameters')
        return pd.DatetimeIndex([])

    # Round sensing times
    cslc_df['stopTime'] = pd.to_datetime(cslc_df['stopTime'])
    cslc_df['time'] = cslc_df['stopTime'].dt.round(f'{round_min}min')

    # Count bursts per path and time
    burst_counts = cslc_df.groupby(['pathNumber', 'time']).size().reset_index(name='count')

    # Calculate average burst count per path
    avg_bursts_per_path = burst_counts.groupby('pathNumber')['count'].mean()

    # Select path with maximum average burst count
    optimal_path = avg_bursts_per_path.idxmax()
    max_avg_count = avg_bursts_per_path.max()

    print(f'Selected path {optimal_path} with average {max_avg_count:.1f} bursts per acquisition')
    print(f'Total paths available: {len(avg_bursts_per_path)}')

    # Get unique times for the optimal path
    path_data = cslc_df[cslc_df['pathNumber'] == optimal_path]
    unique_times = sorted(path_data['time'].unique())

    print(f'Found {len(unique_times)} unique acquisition times')

    return pd.DatetimeIndex(unique_times)
```
